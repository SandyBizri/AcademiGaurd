## Rule-Based Risk Scoring

We define a simple risk scoring system based on:

- G3 < 10 → +3 points
- failures ≥ 1 → +3 points
- studytime ≤ 1 → +2 points
- absences > 10 → +2 points

Higher total score = Higher academic risk.

```python
def calculate_risk(row):
    score = 0

    if row["G3"] < 10:
        score += 3
    if row["failures"] >= 1:
        score += 3
    if row["studytime"] <= 1:
        score += 2
    if row["absences"] > 10:
        score += 2

    return score

df["risk_score"] = df.apply(calculate_risk, axis=1)
```

```python
def classify_risk(score):
    if score >= 7:
        return "High"
    elif score >= 4:
        return "Medium"
    else:
        return "Low"

df["risk_level"] = df["risk_score"].apply(classify_risk)

df.head()
```

## Risk Level Distribution

We now visualize how students are distributed across:

- Low Risk
- Medium Risk
- High Risk

```python
plt.figure()
sns.countplot(data=df, x="risk_level")
plt.title("Risk Level Distribution")
plt.xlabel("Risk Level")
plt.ylabel("Number of Students")
plt.show()
```

## Key Findings

We calculate:

1. Percentage of High Risk students
2. Average final grade for High Risk students
3. Relationship between failures and risk

```python
high_risk = df[df["risk_level"] == "High"]

print("High Risk Percentage:",
      round(len(high_risk)/len(df)*100, 2), "%")

print("Average G3 for High Risk:",
      round(high_risk["G3"].mean(), 2))

print("Average Failures for High Risk:",
      round(high_risk["failures"].mean(), 2))
```
